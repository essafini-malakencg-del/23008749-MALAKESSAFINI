```python
# Sources réelles intégrées à scikit-learn : trois problèmes différents
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.datasets import load_diabetes, load_breast_cancer, load_wine
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score, f1_score, roc_auc_score, silhouette_score
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import make_pipeline
GRAINE = 42

jeu_regression = load_diabetes()
X_reg, y_reg = jeu_regression.data, jeu_regression.target
jeu_classification = load_breast_cancer()
X_cls, y_cls = jeu_classification.data, jeu_classification.target
jeu_non_supervise = load_wine()
X_clu = StandardScaler().fit_transform(jeu_non_supervise.data)  # Exploration sans cible

Xr_ent, Xr_test, yr_ent, yr_test = train_test_split(X_reg, y_reg, test_size=.25, random_state=GRAINE)
Xc_ent, Xc_test, yc_ent, yc_test = train_test_split(X_cls, y_cls, test_size=.25, stratify=y_cls, random_state=GRAINE)
print('Diabetes:', X_reg.shape, '| Breast Cancer:', X_cls.shape, '| Wine:', X_clu.shape)
print('Ces données sont réelles ; les étiquettes du vin ne sont pas employées dans le clustering.')
```

### Outils communs de validation
Ces fonctions réalisent la validation croisée sur l'**entraînement** uniquement, afin de ne jamais régler un modèle avec le test.

```python
# Outils communs : tout réglage se fait par validation croisée sur l'ENTRAÎNEMENT uniquement
import time, warnings
from sklearn.model_selection import cross_val_score, KFold, StratifiedKFold
warnings.filterwarnings('ignore')
cv_r = KFold(5, shuffle=True, random_state=GRAINE)
cv_c = StratifiedKFold(5, shuffle=True, random_state=GRAINE)
def cv_mae(m): return -cross_val_score(m, Xr_ent, yr_ent, cv=cv_r, scoring='neg_mean_absolute_error').mean()
def cv_r2(m):  return cross_val_score(m, Xr_ent, yr_ent, cv=cv_r, scoring='r2').mean()
def cv_f1(m):  return cross_val_score(m, Xc_ent, yc_ent, cv=cv_c, scoring='f1').mean()
def cv_auc(m): return cross_val_score(m, Xc_ent, yc_ent, cv=cv_c, scoring='roc_auc').mean()
print('Outils de validation croisée prêts (5 plis, graine', GRAINE, ')')
```

## 1. Régression linéaire

**Variante C — question spécifique :** Quel diagnostic du résidu indique une mauvaise spécification linéaire ?

**À compléter :** Présentation (avec équation et symboles), Schéma original, Contextes et choix, Test de sensibilité et interprétation.

> Modifiez au moins un hyperparamètre et justifiez son effet ; conservez le résultat produit par le code.


```python
from sklearn.linear_model import LinearRegression
modele = LinearRegression().fit(Xr_ent, yr_ent)
predictions = modele.predict(Xr_test)
print('MAE régression linéaire :', round(mean_absolute_error(yr_test, predictions), 2))
print('R² :', round(r2_score(yr_test, predictions), 3))
```

### Présentation
**Catégorie :** apprentissage supervisé, régression. **Entrées :** variables numériques ; **sortie :** une valeur continue (progression de la maladie). **Hypothèse clé :** relation linéaire entre variables et cible, bruit centré de variance constante, variables peu colinéaires. **Étapes :** (1) poser le modèle linéaire, (2) mesurer l'erreur quadratique sur l'entraînement, (3) la minimiser (moindres carrés, solution exacte), (4) prédire sur de nouvelles observations.

$$\hat y=\beta_0+\sum_{j=1}^{p}\beta_j x_j,\qquad \hat\beta=\arg\min_\beta\sum_{i=1}^{n}(y_i-\hat y_i)^2$$

$\hat y$ : prédiction ; $x_j$ : variable $j$ ; $p$ : nombre de variables ; $n$ : nombre d'observations ; $\beta_0$ : ordonnée à l'origine ; $\beta_j$ : coefficient de la variable $j$. **Paramètres appris :** les $\beta$. **Hyperparamètre :** `fit_intercept` (fixé par nous, non appris).

### Schéma original
**À dessiner vous-même et à insérer ici (photo ou figure) :** Trois cadres : *initialisation* (nuage de points + droite quelconque, segments d'erreur longs) → *apprentissage* (la droite pivote jusqu'à minimiser la somme des carrés des segments) → *sortie* (droite finale + résidus $y-\hat y$). Légende : choix illustré = fonction objectif des moindres carrés.

*Conseil : légendez chaque étape et écrivez sous le dessin le choix algorithmique illustré.*

```python
# --- C4 : diagnostic des résidus + test de sensibilité (algorithme 1) ---
residus = yr_test - predictions
print('RMSE :', round(mean_squared_error(yr_test, predictions)**.5, 2))
print('Résidu moyen :', round(residus.mean(), 2), '| écart-type :', round(residus.std(), 2))
plt.scatter(predictions, residus, alpha=.7); plt.axhline(0, color='red')
plt.xlabel('Prédiction'); plt.ylabel('Résidu'); plt.title('Résidus vs prédictions — régression linéaire'); plt.show()
tercile = pd.qcut(predictions, 3, labels=['bas', 'moyen', 'haut'])
print(pd.Series(residus).groupby(np.asarray(tercile)).std().round(1).to_string())
for fi in [True, False]:
    m = LinearRegression(fit_intercept=fi).fit(Xr_ent, yr_ent)
    print('fit_intercept =', fi, '| MAE test :', round(mean_absolute_error(yr_test, m.predict(Xr_test)), 2),
          '| R² test :', round(r2_score(yr_test, m.predict(Xr_test)), 3))
print('R² test selon 5 séparations différentes :',
      [round(r2_score(d, LinearRegression().fit(a, c).predict(b)), 3)
       for a, b, c, d in (train_test_split(X_reg, y_reg, test_size=.25, random_state=s) for s in range(5))])
```

### Contextes et choix
- **Finance :** estimer le coût d'un sinistre ou le prix d'un bien à partir de variables numériques. **Marketing :** prévoir les ventes selon le budget publicitaire.
- **Données :** tableau numérique, cible continue, peu de colinéarité. **Concurrent :** Ridge (plus stable si variables corrélées).
- **Mal adapté :** relation non linéaire, variance des erreurs croissante, forte colinéarité.
- **Métrique :** MAE (erreur lisible dans l'unité de la cible), RMSE si les grosses erreurs coûtent cher, R² pour se comparer à la moyenne.

### Test de sensibilité et interprétation
**Résultats du code :** MAE test 41,55 ; RMSE 53,37 ; R² 0,485. Écart-type des résidus par tercile de prédiction : 34,6 (bas), 57,5 (moyen), 64,9 (haut). Avec `fit_intercept=False` : MAE 156,36 et R² = −3,94. R² test sur 5 séparations : de 0,359 à 0,467.

**Interprétation.** Le diagnostic d'une mauvaise spécification est un graphique résidus vs prédictions qui montre une **courbure** (non-linéarité) ou un **entonnoir** (variance non constante). Ici les résidus s'élargissent avec la prédiction : signe d'hétéroscédasticité. Sans ordonnée à l'origine, le modèle est forcé de passer par 0 alors que la cible a une moyenne d'environ 150 : il devient pire que la moyenne (R² < 0).

**Ce qu'on ne peut pas conclure :** avec une seule séparation, le R² varie de 0,36 à 0,47 selon la graine ; « 0,485 » n'est donc pas *la* performance du modèle.

## 2. Ridge

**Variante C — question spécifique :** Comment évoluent les coefficients quand alpha augmente ?

**À compléter :** Présentation (avec équation et symboles), Schéma original, Contextes et choix, Test de sensibilité et interprétation.

> Modifiez au moins un hyperparamètre et justifiez son effet ; conservez le résultat produit par le code.


```python
from sklearn.linear_model import Ridge
modele = make_pipeline(StandardScaler(), Ridge(alpha=12.0)).fit(Xr_ent, yr_ent)
print('MAE Ridge :', round(mean_absolute_error(yr_test, modele.predict(Xr_test)), 2))
```

### Présentation
**Catégorie :** supervisé, régression linéaire régularisée (pénalité L2). **Entrées/sortie :** comme la régression linéaire. **Hypothèse clé :** les coefficients doivent rester de petite taille ; utile lorsque les variables sont corrélées. **Étapes :** standardiser les variables (la pénalité dépend de l'échelle), minimiser l'erreur quadratique + pénalité, prédire.

$$\hat\beta=\arg\min_\beta\ \sum_{i=1}^{n}(y_i-\beta_0-x_i^\top\beta)^2+\alpha\,\|\beta\|_2^2$$

$\alpha\ge 0$ : force de la pénalité ; $\|\beta\|_2^2=\sum_j\beta_j^2$ ; les autres symboles sont ceux de la régression linéaire. **Paramètres appris :** $\beta$. **Hyperparamètre :** $\alpha$ (choisi par validation croisée, pas appris par l'optimisation).

### Schéma original
**À dessiner vous-même et à insérer ici (photo ou figure) :** Un schéma à deux cercles de contrainte : la zone d'erreur (ellipses centrées sur $\hat\beta_{MCO}$) et le disque $\|\beta\|_2\le r$ ; flèche montrant que la solution Ridge est le point de contact, plus proche de l'origine quand $\alpha$ augmente. Légende : choix illustré = pénalité L2.

*Conseil : légendez chaque étape et écrivez sous le dessin le choix algorithmique illustré.*

```python
# --- C4 : effet de alpha sur les coefficients (Ridge) ---
from sklearn.linear_model import Ridge
alphas = [0.01, 1, 12, 100, 1000]
chemin = []
print('alpha | norme des coefficients | MAE CV')
for a in alphas:
    pipe = make_pipeline(StandardScaler(), Ridge(alpha=a))
    pipe.fit(Xr_ent, yr_ent); chemin.append(pipe[-1].coef_)
    print(f'{a:>7} | {np.linalg.norm(pipe[-1].coef_):8.2f} | {cv_mae(pipe):.2f}')
plt.plot(np.log10(alphas), np.array(chemin)); plt.xlabel('log10(alpha)'); plt.ylabel('coefficient')
plt.title('Ridge : trajectoire des coefficients'); plt.show()
```

### Contextes et choix
- **Finance :** modèle de score avec nombreux indicateurs corrélés (ratios comptables). **Marketing :** réponse aux campagnes avec canaux publicitaires corrélés.
- **Données :** tableau numérique standardisé, colinéarité possible. **Concurrent :** Lasso (sélection de variables) ou régression linéaire simple.
- **Mal adapté :** si l'on veut éliminer des variables (Ridge ne les annule jamais) ou si la relation est non linéaire.
- **Métrique :** MAE/RMSE sur validation croisée pour choisir $\alpha$.

### Test de sensibilité et interprétation
**Résultats du code (α = 0,01 / 1 / 12 / 100 / 1000) :** norme des coefficients 70,31 / 59,75 / 41,99 / 33,79 / 15,16 ; MAE en validation croisée 45,81 / 45,75 / 45,91 / 47,19 / 56,01.

**Interprétation.** Quand α augmente, tous les coefficients **rétrécissent progressivement sans devenir nuls** (voir la trajectoire). Pour α ≤ 12 la MAE ne change presque pas (écarts de l'ordre de 0,1–0,2) ; au-delà, la pénalité écrase trop le signal : sous-ajustement (MAE 56,01 à α = 1000). La valeur retenue par le code de base (α = 12) n'est donc pas distinguable de α = 1.

**Limite :** les différences entre 0,01, 1 et 12 sont plus petites que la variabilité probable entre plis ; on ne peut pas conclure qu'un de ces α est meilleur.

## 3. Lasso

**Variante C — question spécifique :** Combien de variables sont annulées pour trois valeurs d’alpha ?

**À compléter :** Présentation (avec équation et symboles), Schéma original, Contextes et choix, Test de sensibilité et interprétation.

> Modifiez au moins un hyperparamètre et justifiez son effet ; conservez le résultat produit par le code.


```python
from sklearn.linear_model import Lasso
modele = make_pipeline(StandardScaler(), Lasso(alpha=2.0, max_iter=15000)).fit(Xr_ent, yr_ent)
print('Nombre de coefficients non nuls :', np.count_nonzero(modele[-1].coef_))
print('MAE Lasso :', round(mean_absolute_error(yr_test, modele.predict(Xr_test)), 2))
```

### Présentation
**Catégorie :** supervisé, régression avec pénalité L1. **Hypothèse clé :** seule une partie des variables compte (modèle parcimonieux). **Étapes :** standardiser, minimiser l'erreur + pénalité L1 (algorithme de descente par coordonnées), certains coefficients tombent exactement à 0, prédire.

$$\hat\beta=\arg\min_\beta\ \frac{1}{2n}\sum_{i=1}^{n}(y_i-\beta_0-x_i^\top\beta)^2+\alpha\sum_{j=1}^{p}|\beta_j|$$

$\alpha$ : force de la pénalité ; $|\beta_j|$ : valeur absolue du coefficient ; autres symboles comme avant. **Paramètres appris :** $\beta$ (dont plusieurs nuls). **Hyperparamètre :** $\alpha$ (et `max_iter`, critère d'arrêt).

### Schéma original
**À dessiner vous-même et à insérer ici (photo ou figure) :** Même schéma que Ridge mais avec un **losange** ($\|\beta\|_1\le r$) : le point de contact tombe sur un sommet, donc sur un axe (coefficient = 0). Montrer 3 positions de plus en plus proches de l'origine pour $\alpha$ croissant. Choix illustré = pénalité L1.

*Conseil : légendez chaque étape et écrivez sous le dessin le choix algorithmique illustré.*

```python
# --- C4 : variables annulées selon alpha (Lasso) ---
from sklearn.linear_model import Lasso
noms = np.array(jeu_regression.feature_names)
for a in [0.5, 2, 10]:
    pipe = make_pipeline(StandardScaler(), Lasso(alpha=a, max_iter=15000)).fit(Xr_ent, yr_ent)
    coef = pipe[-1].coef_
    print(f'alpha={a:>4} | non nuls : {np.count_nonzero(coef)}/10 | annulées : {list(map(str, noms[coef == 0]))} | MAE CV : {cv_mae(make_pipeline(StandardScaler(), Lasso(alpha=a, max_iter=15000))):.2f}')
```

### Contextes et choix
- **Commerce international :** repérer parmi de nombreux indicateurs pays (tarifs, PIB, distance, langue…) ceux qui expliquent un volume d'échanges. **Marketing :** choisir les variables clients réellement utiles.
- **Données :** nombreuses variables, peu d'observations. **Concurrent :** Ridge ou Elastic Net.
- **Mal adapté :** variables très corrélées (le Lasso en garde une presque au hasard) ; sélection alors instable.
- **Métrique :** MAE en validation croisée + nombre de variables conservées (lisibilité).

### Test de sensibilité et interprétation
**Résultats du code :** α = 0,5 → 9/10 coefficients non nuls (annulée : s2), MAE CV 45,93 ; α = 2 → 8/10 (s2, s4), MAE CV 46,54 ; α = 10 → 4/10 (annulées : age, sex, s1, s2, s4, s6), MAE CV 49,67.

**Interprétation.** Plus α est grand, plus de variables sont éliminées : on gagne en lisibilité mais on perd en précision (passer de 9 à 4 variables coûte environ 3,7 points de MAE CV). Le Lasso fait donc de la **sélection de variables** que Ridge ne fait pas.

**Limite :** s1 et s2 (mesures de cholestérol) sont fortement corrélées ; le fait que s2 soit éliminée n'indique pas qu'elle est inutile, mais qu'elle est redondante avec une autre. Une seule séparation de données ne garantit pas la stabilité de la sélection.

## 4. Régression polynomiale

**Variante C — question spécifique :** Le degré 2 améliore-t-il le test ou seulement l’entraînement ?

**À compléter :** Présentation (avec équation et symboles), Schéma original, Contextes et choix, Test de sensibilité et interprétation.

> Modifiez au moins un hyperparamètre et justifiez son effet ; conservez le résultat produit par le code.


```python
from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import Ridge
modele = make_pipeline(StandardScaler(), PolynomialFeatures(degree=2, include_bias=False), Ridge(alpha=100.)).fit(Xr_ent, yr_ent)
print('R² polynomial :', round(r2_score(yr_test, modele.predict(Xr_test)), 3))
```

### Présentation
**Catégorie :** supervisé, régression non linéaire en les variables mais linéaire en les coefficients. **Hypothèse clé :** la cible dépend de puissances et de produits des variables. **Étapes :** standardiser, créer les termes de degré $d$ (carrés, produits), ajuster une régression pénalisée (Ridge), prédire.

$$\hat y=\beta_0+\sum_j\beta_jx_j+\sum_{j\le k}\beta_{jk}x_jx_k\quad(d=2)$$

$x_jx_k$ : terme d'interaction (ou carré si $j=k$) ; $\beta_{jk}$ : son coefficient. Avec 10 variables, le degré 2 donne 65 termes, le degré 3 donne 285. **Paramètres appris :** les $\beta$. **Hyperparamètres :** degré $d$, $\alpha$ de Ridge.

### Schéma original
**À dessiner vous-même et à insérer ici (photo ou figure) :** Trois courbes sur le même nuage : droite (degré 1), parabole (degré 2), courbe très ondulée passant par tous les points (degré 3 sans pénalité). Légende : capacité croissante, risque de surajustement. Choix illustré = base de fonctions.

*Conseil : légendez chaque étape et écrivez sous le dessin le choix algorithmique illustré.*

```python
# --- C4 : degré polynomial, entraînement vs validation ---
for d in [1, 2, 3]:
    pipe = make_pipeline(StandardScaler(), PolynomialFeatures(degree=d, include_bias=False), Ridge(alpha=100.))
    pipe.fit(Xr_ent, yr_ent)
    print(f'degré {d} | variables : {pipe[1].n_output_features_:>3} | R² entraînement : {r2_score(yr_ent, pipe.predict(Xr_ent)):.3f}'
          f' | R² validation croisée : {cv_r2(pipe):.3f} | R² test (lecture finale) : {r2_score(yr_test, pipe.predict(Xr_test)):.3f}')
```

### Contextes et choix
- **Logistique/gestion :** coût de transport selon distance et charge (effets en courbe). **Finance :** sensibilité non linéaire d'un prix à un taux.
- **Données :** peu de variables, courbure visible. **Concurrent :** forêt ou boosting (sans choisir les termes à la main).
- **Mal adapté :** beaucoup de variables (explosion du nombre de termes), peu d'observations.
- **Métrique :** R² ou MAE en **validation croisée** (le R² d'entraînement augmente toujours avec le degré).

**Résultats du code (degré 1 / 2 / 3) :** nombre de variables 10 / 65 / 285 ; R² entraînement 0,502 / 0,557 / 0,662 ; R² validation croisée 0,448 / 0,416 / 0,222 ; R² test (lecture finale seulement) 0,490 / 0,501 / 0,368.

**Interprétation.** Le degré 2 améliore nettement l'entraînement (+0,055) mais **dégrade la validation croisée** (−0,032) ; le gain de 0,011 sur le test est trop faible pour être fiable. Le degré 3 surajuste clairement (0,662 à l'entraînement contre 0,368 au test). Sur ces données, le choix raisonnable, fait par validation croisée, est le degré 1.

## 6. Forêt aléatoire — régression

**Variante C — question spécifique :** La forêt réduit-elle la variance par rapport à l’arbre ?

**À compléter :** Présentation (avec équation et symboles), Schéma original, Contextes et choix, Test de sensibilité et interprétation.

> Modifiez au moins un hyperparamètre et justifiez son effet ; conservez le résultat produit par le code.


```python
from sklearn.ensemble import RandomForestRegressor
modele = RandomForestRegressor(n_estimators=60, max_depth=6, n_jobs=-1, random_state=GRAINE).fit(Xr_ent, yr_ent)
print('MAE forêt :', round(mean_absolute_error(yr_test, modele.predict(Xr_test)), 2))
```

### Présentation
**Catégorie :** supervisé, régression par ensemble (bagging). **Hypothèse clé :** moyenner de nombreux arbres différents réduit la variance sans augmenter beaucoup le biais. **Étapes :** (1) tirer $B$ échantillons bootstrap, (2) entraîner un arbre par échantillon avec un sous-ensemble aléatoire de variables à chaque coupure, (3) moyenner les prédictions.

$$\hat y(x)=\frac{1}{B}\sum_{b=1}^{B}T_b(x)$$

$B$ : nombre d'arbres (`n_estimators`) ; $T_b$ : $b$-ième arbre. **Paramètres appris :** structure de chaque arbre. **Hyperparamètres :** $B$, `max_depth`, `max_features`.

### Schéma original
**À dessiner vous-même et à insérer ici (photo ou figure) :** Trois petits arbres un peu différents (échantillons bootstrap) avec leurs prédictions (p. ex. 120, 160, 140) qui convergent vers une flèche « moyenne = 140 ». Légende : initialisation = échantillons aléatoires ; sortie = moyenne. Choix illustré = agrégation.

*Conseil : légendez chaque étape et écrivez sous le dessin le choix algorithmique illustré.*

```python
# --- C4 : la forêt réduit-elle la variance ? ---
from sklearn.model_selection import cross_val_score
def mae_par_pli(m): return -cross_val_score(m, Xr_ent, yr_ent, cv=cv_r, scoring='neg_mean_absolute_error')
arbre = mae_par_pli(DecisionTreeRegressor(max_depth=6, random_state=GRAINE))
foret = mae_par_pli(RandomForestRegressor(n_estimators=60, max_depth=6, random_state=GRAINE, n_jobs=-1))
print('Arbre  (profondeur 6) : MAE moyenne', arbre.mean().round(2), '| écart-type entre plis', arbre.std().round(2))
print('Forêt  (60 arbres)    : MAE moyenne', foret.mean().round(2), '| écart-type entre plis', foret.std().round(2))
for n in [5, 20, 60, 200]:
    scores = [-cross_val_score(RandomForestRegressor(n_estimators=n, max_depth=6, random_state=s, n_jobs=-1), Xr_ent, yr_ent,
              cv=cv_r, scoring='neg_mean_absolute_error').mean() for s in range(5)]
    print(f'n_estimators={n:>3} | MAE CV moyenne {np.mean(scores):.2f} | écart-type selon la graine {np.std(scores):.3f}')
```

### Contextes et choix
- **Finance :** estimer une valeur immobilière ou un risque avec interactions inconnues. **Logistique :** prévoir un délai de livraison.
- **Données :** tableau mixte, non-linéarités, interactions. **Concurrent :** boosting (souvent plus précis, plus sensible au réglage).
- **Mal adapté :** extrapolation hors de la plage observée ; besoin d'un modèle très interprétable.
- **Métrique :** MAE/RMSE en validation croisée, avec l'écart-type entre plis comme mesure de stabilité.

### Test de sensibilité et interprétation
**Résultats du code :** arbre (profondeur 6) MAE CV 56,71 avec écart-type entre plis 6,03 ; forêt de 60 arbres MAE CV 50,22 avec écart-type 2,82. Selon `n_estimators` (5 / 20 / 60 / 200) : MAE CV 52,36 / 50,51 / 49,92 / 49,68 et écart-type selon la graine 0,490 / 0,377 / 0,207 / 0,223.

**Interprétation.** Oui, la forêt réduit la variance : l'écart-type entre plis est divisé par plus de 2 et l'erreur baisse d'environ 6,5 points. Le gain en nombre d'arbres est rapide jusqu'à ~60 puis négligeable (49,92 → 49,68), la stabilité se stabilise aussi. Le coût de calcul, lui, croît avec $B$.

**Limite :** l'écart-type entre plis mélange la variabilité des données et celle de l'algorithme ; 5 graines donnent une estimation grossière.

### Présentation
**Catégorie :** supervisé, régression par boosting de gradient (version histogrammes). **Hypothèse clé :** corriger séquentiellement les erreurs d'un modèle par de petits arbres. **Étapes :** initialiser par la moyenne, calculer les résidus (gradients de la perte), entraîner un petit arbre dessus, l'ajouter avec un pas $\nu$, répéter `max_iter` fois.

$$F_m(x)=F_{m-1}(x)+\nu\,h_m(x),\qquad h_m\approx r_i=-\frac{\partial L(y_i,F_{m-1}(x_i))}{\partial F_{m-1}(x_i)}$$

$F_m$ : modèle après $m$ arbres ; $h_m$ : $m$-ième petit arbre ; $\nu$ : taux d'apprentissage ; $L$ : perte ; $r_i$ : résidu de l'observation $i$. **Paramètres appris :** les arbres. **Hyperparamètres :** `max_iter`, `learning_rate`, `max_leaf_nodes`.

Choix illustré = descente de gradient dans l'espace des fonctions.

*Conseil : légendez chaque étape et écrivez sous le dessin le choix algorithmique illustré.*

```python
# --- C4 : max_iter (boosting de régression) : coût et surajustement ---
print('max_iter | temps (s) | MAE entraînement | MAE validation croisée')
for it in [20, 100, 300, 800]:
    t0 = time.perf_counter()
    gb = HistGradientBoostingRegressor(max_iter=it, max_leaf_nodes=9, random_state=GRAINE).fit(Xr_ent, yr_ent)
    t = time.perf_counter() - t0
    print(f'{it:>8} | {t:9.2f} | {mean_absolute_error(yr_ent, gb.predict(Xr_ent)):16.2f} | {cv_mae(HistGradientBoostingRegressor(max_iter=it, max_leaf_nodes=9, random_state=GRAINE)):.2f}')
```

### Contextes et choix
- **Finance :** scoring de montants (pertes, revenus) sur données tabulaires. **Commerce :** prévision de demande avec nombreuses variables.
- **Données :** tabulaires, tailles moyennes à grandes. **Concurrent :** forêt (moins de réglage).
- **Mal adapté :** très petits échantillons (surajustement rapide), besoin d'explications simples.

### Test de sensibilité et interprétation

**Interprétation.** Plus on ajoute d'itérations, plus l'entraînement est ajusté (MAE 1,62) mais plus la validation se dégrade : **surajustement**. ## 8. Régression logistique

**Variante C — question spécifique :** Quel effet de C sur les probabilités et la frontière ?

**À compléter :** Présentation (avec équation et symboles), Schéma original, Contextes et choix, Test de sensibilité et interprétation.

```python
from sklearn.linear_model import LogisticRegression
modele = make_pipeline(StandardScaler(), LogisticRegression(C=.5, max_iter=2000, random_state=GRAINE)).fit(Xc_ent, yc_ent)
print('F1 logistique :', round(f1_score(yc_test, modele.predict(Xc_test)), 3))
```

### Présentation
**Catégorie :** supervisé, classification binaire (modèle linéaire probabiliste). **Entrées :** variables numériques ; **sortie :** probabilité de la classe 1 puis étiquette. **Hypothèse clé :** le log-odds est linéaire en les variables. **Étapes :** combinaison linéaire $z$, transformation sigmoïde, minimisation de la log-vraisemblance pénalisée, seuil à 0,5.

$$P(y=1\mid x)=\sigma(z)=\frac{1}{1+e^{-z}},\ z=\beta_0+x^\top\beta;\qquad \min_\beta\ -\sum_i\big[y_i\ln p_i+(1-y_i)\ln(1-p_i)\big]+\frac{1}{2C}\|\beta\|_2^2$$

$\sigma$ : fonction sigmoïde ; $p_i=P(y_i=1\mid x_i)$ ; $C$ : inverse de la force de régularisation. **Paramètres appris :** $\beta$. **Hyperparamètre :** $C$. (Convention scikit-learn sur Breast Cancer : classe 0 = malin, classe 1 = bénin.)

### Schéma original

Trois colonnes : initialisation (droite aléatoire, probabilités ≈ 0,5), apprentissage (la droite se tourne), sortie (probabilités proches de 0 ou 1). Choix illustré = fonction objectif (log-vraisemblance).

*Conseil : légendez chaque étape et écrivez sous le dessin le choix algorithmique illustré.*

```python
# --- C4 : effet de C (régression logistique) ---
print('C | norme des coefficients | confiance moyenne (|p-0.5|*2) | F1 CV | AUC CV')
for C in [0.01, 0.5, 10, 100]:
    pipe = make_pipeline(StandardScaler(), LogisticRegression(C=C, max_iter=5000, random_state=GRAINE)).fit(Xc_ent, yc_ent)
    p = pipe.predict_proba(Xc_ent)[:, 1]
    print(f'{C:>6} | {np.linalg.norm(pipe[-1].coef_):8.2f} | {np.mean(np.abs(p-.5)*2):.3f} | {cv_f1(pipe):.3f} | {cv_auc(pipe):.3f}')
```

### Contextes et choix
- **Finance :** probabilité de défaut de paiement. **Marketing :** probabilité d'achat ou de désabonnement.
- **Données :** variables numériques standardisées, cible binaire. **Concurrent :** SVM, forêt.
- **Mal adapté :** frontière très non linéaire, classes parfaitement séparables sans régularisation.
- **Métrique :** F1 et ROC-AUC (classes déséquilibrées) ; en médical ou en risque, regarder aussi le type d'erreur (faux négatifs).

### Test de sensibilité et interprétation
**Résultats du code (C = 0,01 / 0,5 / 10 / 100) :** norme des coefficients 0,81 / 2,87 / 8,38 / 27,82 ; confiance moyenne des probabilités 0,736 / 0,930 / 0,961 / 0,975 ; F1 CV 0,963 / 0,982 / 0,974 / 0,972 ; AUC CV 0,993 / 0,996 / 0,990 / 0,988.

## 10. Naive Bayes

**Variante C — question spécifique :** Dans quel cas l’hypothèse d’indépendance devient-elle critique ?

**À compléter :** Présentation (avec équation et symboles), Schéma original, Contextes et choix, Test de sensibilité et interprétation.

> Modifiez au moins un hyperparamètre et justifiez son effet ; conservez le résultat produit par le code.


```python
from sklearn.naive_bayes import GaussianNB
modele = GaussianNB(var_smoothing=1e-8).fit(Xc_ent, yc_ent)
print('F1 Naive Bayes :', round(f1_score(yc_test, modele.predict(Xc_test)), 3))
```

### Présentation
**Catégorie :** supervisé, classification probabiliste (modèle génératif). **Hypothèse clé :** variables **indépendantes conditionnellement à la classe**, de loi gaussienne. **Étapes :** estimer pour chaque classe la proportion et la moyenne/variance de chaque variable, calculer la probabilité a posteriori par le théorème de Bayes, choisir la classe la plus probable.

$$\hat y=\arg\max_c\ P(c)\prod_{j=1}^{p}\mathcal N\!\left(x_j;\mu_{cj},\sigma_{cj}^2\right)$$

$P(c)$ : proportion de la classe $c$ ; $\mu_{cj},\sigma_{cj}^2$ : moyenne et variance de la variable $j$ dans la classe $c$. **Paramètres appris :** $P(c),\mu_{cj},\sigma^2_{cj}$. **Hyperparamètre :** `var_smoothing` (fraction de la plus grande variance ajoutée pour la stabilité numérique).

### Schéma original
**À dessiner vous-même et à insérer ici (photo ou figure) :** Pour une variable, deux courbes en cloche (une par classe) et une observation marquée ; en dessous, les probabilités de chaque variable multipliées par un symbole « × » pour donner la probabilité finale. Légende : l'indépendance permet de multiplier. Choix illustré = factorisation de la vraisemblance.

*Conseil : légendez chaque étape et écrivez sous le dessin le choix algorithmique illustré.*

```python
# --- C4 : indépendance conditionnelle et var_smoothing ---
corr = np.abs(np.corrcoef(Xc_ent, rowvar=False)); np.fill_diagonal(corr, 0)
print('Paires de variables avec |corrélation| > 0.9 :', int((np.triu(corr) > .9).sum()), 'sur', 30*29//2)
for vs in [1e-12, 1e-8, 1e-4, 1e-2, 1]:
    print(f'var_smoothing={vs:g} | F1 CV : {cv_f1(GaussianNB(var_smoothing=vs)):.3f}')
print('Référence régression logistique | F1 CV :', round(cv_f1(make_pipeline(StandardScaler(), LogisticRegression(max_iter=5000))), 3))
```

### Contextes et choix
- **Marketing :** classer des avis ou courriels (variante multinomiale). **Gestion :** triage rapide avec peu de données.
- **Données :** petit échantillon, beaucoup de variables, apprentissage très rapide. **Concurrent :** régression logistique.
- **Mal adapté :** variables fortement corrélées entre elles (preuves comptées plusieurs fois, probabilités trop extrêmes).
- **Métrique :** F1 et AUC ; la calibration des probabilités est mauvaise en général.

### Test de sensibilité et interprétation
**Résultats du code :** 21 paires de variables sur 435 ont une corrélation absolue > 0,9. F1 CV selon `var_smoothing` (1e-12 / 1e-8 / 1e-4 / 1e-2 / 1) : 0,946 / 0,947 / 0,939 / 0,929 / 0,875. Régression logistique de référence : 0,983.

**Interprétation.** L'hypothèse d'indépendance devient **critique** quand beaucoup de variables sont redondantes : ici rayon, périmètre et aire d'un noyau cellulaire sont presque la même information, comptée plusieurs fois dans le produit. Cela explique en partie l'écart avec la logistique (0,947 contre 0,983). Un `var_smoothing` trop élevé (1) écrase les variances et fait perdre de l'information (F1 0,875).

**Limite :** l'écart de F1 peut aussi venir d'autres causes (forme non gaussienne) ; on ne peut pas l'attribuer uniquement à la corrélation sans une expérience dédiée (par exemple retirer les variables redondantes).

$w,b$ : hyperplan ; $\xi_i$ : variable de relâchement (erreur tolérée) ; $y_i\in\{-1,+1\}$ ; $C$ : pénalité des erreurs ; $\gamma$ : portée du noyau RBF. **Paramètres appris :** $w,b$ (ou coefficients des vecteurs supports). **Hyperparamètres :** noyau, $C$, $\gamma$.

**Interprétation et choix.** Le noyau linéaire est aussi performant que le RBF (écarts de l'ordre de 0,002 à 0,009) avec un modèle plus simple et interprétable : c'est le choix justifié ici (30 variables standardisées, classes quasi séparables linéairement). $F_m$ : score après $m$ arbres ; $h_m$ : arbre ajusté sur le gradient négatif de $L$ ; $\nu$ : taux d'apprentissage ; $\sigma$ : sigmoïde. **Paramètres appris :** les arbres. **Hyperparamètres :** `max_iter`, `learning_rate`, `max_leaf_nodes`.

## 15. AdaBoost

**Variante C — question spécifique :** Que change learning_rate quand n_estimators reste fixe ?

**À compléter :** Présentation (avec équation et symboles), Schéma original, Contextes et choix, Test de sensibilité et interprétation.

> Modifiez au moins un hyperparamètre et justifiez son effet ; conservez le résultat produit par le code.


```python
from sklearn.ensemble import AdaBoostClassifier
modele = AdaBoostClassifier(n_estimators=50, learning_rate=.6, random_state=GRAINE).fit(Xc_ent, yc_ent)
print('F1 AdaBoost :', round(f1_score(yc_test, modele.predict(Xc_test)), 3))
```

### Présentation
**Catégorie :** supervisé, classification par boosting adaptatif. **Hypothèse clé :** combiner des classifieurs faibles (souches) en réaffectant du poids aux erreurs. **Étapes :** poids égaux, entraîner un classifieur faible, calculer son erreur pondérée, lui donner un poids, augmenter le poids des observations mal classées, répéter.

$$F(x)=\operatorname{sign}\Big(\sum_{m=1}^{M}\alpha_m h_m(x)\Big),\quad \alpha_m=\nu\ln\frac{1-\varepsilon_m}{\varepsilon_m},\quad w_i\leftarrow w_i\,e^{\alpha_m\mathbf 1[h_m(x_i)\ne y_i]}$$

$h_m$ : classifieur faible $m$ ; $\varepsilon_m$ : son erreur pondérée ; $\alpha_m$ : son poids dans le vote ; $\nu$ : `learning_rate` ; $w_i$ : poids de l'observation $i$. **Paramètres appris :** $h_m,\alpha_m$. **Hyperparamètres :** `n_estimators`, `learning_rate`.

### Schéma original
**À dessiner vous-même et à insérer ici (photo ou figure) :** Trois cadres montrant le même nuage : poids égaux (points de même taille) → après le 1er classifieur, les points mal classés grossissent → après le 2e, nouvelle frontière. Légende : taille = poids. Choix illustré = pondération des erreurs.

*Conseil : légendez chaque étape et écrivez sous le dessin le choix algorithmique illustré.*

```python
# --- C4 : learning_rate avec n_estimators fixe à 50 ---
print('learning_rate | F1 entraînement | F1 CV | AUC CV')
for lr in [.05, .2, .6, 1.0]:
    ada = AdaBoostClassifier(n_estimators=50, learning_rate=lr, random_state=GRAINE)
    print(f'{lr:>13} | {f1_score(yc_ent, ada.fit(Xc_ent, yc_ent).predict(Xc_ent)):15.3f} | {cv_f1(AdaBoostClassifier(n_estimators=50, learning_rate=lr, random_state=GRAINE)):.3f} | {cv_auc(AdaBoostClassifier(n_estimators=50, learning_rate=lr, random_state=GRAINE)):.3f}')
```

### Contextes et choix
- **Finance :** classification de transactions avec peu de variables informatives. **Marketing :** classer des prospects avec règles simples combinées.
- **Données :** tabulaires, bruit modéré. **Concurrent :** boosting de gradient.
- **Mal adapté :** données bruitées ou avec erreurs d'étiquette (le boosting s'acharne sur elles).
- **Métrique :** F1 et AUC.

### Test de sensibilité et interprétation
**Résultats du code (learning_rate = 0,05 / 0,2 / 0,6 / 1,0, n_estimators = 50) :** F1 entraînement 0,972 / 0,987 / 1,000 / 1,000 ; F1 CV 0,959 / 0,968 / 0,974 / 0,974 ; AUC CV 0,981 / 0,987 / 0,993 / 0,994.

**Interprétation.** Avec le nombre d'estimateurs fixé, `learning_rate` règle l'importance donnée à chaque classifieur : un taux faible rend l'apprentissage trop lent (sous-ajustement à 50 estimateurs) ; un taux élevé l'accélère jusqu'au plateau. Il existe un compromis `learning_rate` × `n_estimators` : un faible taux demande plus d'estimateurs.

**Limite :** entre 0,6 et 1,0 on ne distingue plus de différence ; un taux plus élevé pourrait aussi amplifier le bruit sur d'autres données.

## 16. XGBoost

**Variante C — question spécifique :** Que change max_depth sur la capacité du modèle ?

**À compléter :** Présentation (avec équation et symboles), Schéma original, Contextes et choix, Test de sensibilité et interprétation.

> Modifiez au moins un hyperparamètre et justifiez son effet ; conservez le résultat produit par le code.


```python
from xgboost import XGBClassifier
modele = XGBClassifier(n_estimators=65, max_depth=2, learning_rate=.08, random_state=GRAINE, eval_metric='logloss', n_jobs=2).fit(Xc_ent, yc_ent)
print('F1 XGBoost :', round(f1_score(yc_test, modele.predict(Xc_test)), 3))
```

### Présentation
**Catégorie :** supervisé, boosting de gradient régularisé (arbres, approximation de second ordre). **Hypothèse clé :** comme le boosting de gradient, avec pénalité explicite sur la complexité des arbres. **Étapes :** calculer gradient et hessienne de la perte, construire un arbre qui maximise le gain régularisé, ajouter l'arbre avec $\eta$, répéter.

$$\mathcal L=\sum_i \ell(y_i,\hat y_i)+\sum_m\Big[\gamma T_m+\tfrac12\lambda\|w_m\|^2\Big]$$

$\ell$ : perte (log-loss) ; $T_m$ : nombre de feuilles de l'arbre $m$ ; $w_m$ : valeurs de ses feuilles ; $\gamma,\lambda$ : pénalités ; $\eta$ : `learning_rate`. **Paramètres appris :** structure et valeurs des arbres. **Hyperparamètres :** `max_depth`, `n_estimators`, `learning_rate`, $\lambda,\gamma$.

### Schéma original
**À dessiner vous-même et à insérer ici (photo ou figure) :** Un arbre avec, sur chaque nœud, le calcul « gain = gain gauche + gain droit − γ » et un nœud barré (coupé car gain < γ). Légende : la pénalité élague les coupures peu utiles. Choix illustré = fonction objectif régularisée.

*Conseil : légendez chaque étape et écrivez sous le dessin le choix algorithmique illustré.*

```python
# --- C4 : max_depth (XGBoost) et capacité ---
print('max_depth | F1 entraînement | F1 validation croisée')
for d in [1, 2, 4, 8]:
    mk = lambda: XGBClassifier(n_estimators=65, max_depth=d, learning_rate=.08, random_state=GRAINE, eval_metric='logloss', n_jobs=2)
    print(f'{d:>9} | {f1_score(yc_ent, mk().fit(Xc_ent, yc_ent).predict(Xc_ent)):15.3f} | {cv_f1(mk()):.3f}')
```

### Contextes et choix
- **Banque/assurance :** scoring de risque et détection de fraude. **E-commerce :** prédiction de conversion.
- **Données :** tabulaires volumineuses, valeurs manquantes tolérées. **Concurrent :** HistGradientBoosting, forêt.
- **Mal adapté :** images, textes bruts, très petits échantillons.
- **Métrique :** F1, AUC, log-loss.

### Test de sensibilité et interprétation

## 19. DBSCAN

**Variante C — question spécifique :** Expliquez pourquoi eps modifie le nombre de points bruit.

**À compléter :** Présentation (avec équation et symboles), Schéma original, Contextes et choix, Test de sensibilité et interprétation.

> Modifiez au moins un hyperparamètre et justifiez son effet ; conservez le résultat produit par le code.


```python
from sklearn.cluster import DBSCAN
etiquettes = DBSCAN(eps=2.5, min_samples=4).fit_predict(X_clu)
print('Groupes détectés hors bruit :', len(set(etiquettes) - {-1}))
print('Points considérés comme bruit :', np.sum(etiquettes == -1))
```

### Présentation
**Catégorie :** non supervisé, partitionnement par densité. **Hypothèse clé :** les groupes sont des zones denses séparées par des zones peu denses ; les points isolés sont du bruit. **Étapes :** pour chaque point, compter les voisins dans un rayon $\varepsilon$ ; si ≥ `min_samples`, c'est un point **cœur** ; relier les cœurs voisins en groupes, rattacher les points frontières, étiqueter le reste « bruit » (−1).

$$N_\varepsilon(x)=\{x' : \|x-x'\|\le\varepsilon\},\qquad x\text{ cœur}\iff |N_\varepsilon(x)|\ge m_{\min}$$

$\varepsilon$ : rayon de voisinage ; $m_{\min}$ : `min_samples`. **Paramètre appris :** les affectations (pas de centres). **Hyperparamètres :** $\varepsilon$, $m_{\min}$, métrique.

### Schéma original
**À dessiner vous-même et à insérer ici (photo ou figure) :** Un nuage avec des cercles de rayon $\varepsilon$ autour de quelques points : cœurs (pleins), frontières (demi-pleins), bruit (croix). Dessiner ensuite le même nuage avec un $\varepsilon$ plus grand où deux groupes fusionnent. Choix illustré = notion de densité.

*Conseil : légendez chaque étape et écrivez sous le dessin le choix algorithmique illustré.*

```python
# --- C4 : effet de eps sur le bruit ---
print('eps | groupes | points bruit')
for eps in [1.5, 2.0, 2.5, 3.0, 3.5]:
    et = DBSCAN(eps=eps, min_samples=4).fit_predict(X_clu)
    print(f'{eps:>3} | {len(set(et) - {-1}):>7} | {int(np.sum(et == -1))}')
```

### Contextes et choix
- **Logistique :** repérer des zones de forte concentration de livraisons et des points isolés. **Finance :** détecter des transactions atypiques à côté de groupes denses.
- **Données :** faible dimension, densités homogènes. **Concurrent :** K-means (si groupes sphériques).
- **Mal adapté :** grande dimension (dès 13 variables les densités se contrastent peu), densités variables.
- **Métrique :** nombre de groupes, part de bruit ; la silhouette est peu fiable avec du bruit.

### Test de sensibilité et interprétation
**Résultats du code (eps = 1,5 / 2,0 / 2,5 / 3,0 / 3,5, min_samples=4) :** groupes hors bruit 2 / 6 / 1 / 1 / 1 ; points bruit 169 / 66 / 23 / 11 / 6.

**Interprétation.** $\varepsilon$ est le rayon du voisinage : s'il est petit, très peu de points ont assez de voisins, donc la plupart sont classés **bruit** (169 sur 178 pour 1,5) ; s'il est grand, tout se rejoint en **un seul groupe** et le bruit disparaît. Le nombre de groupes n'est pas monotone (6 pour eps = 2,0) : DBSCAN est très sensible à ce paramètre, surtout en 13 dimensions où les densités sont peu contrastées.

**Limite :** il n'existe pas de valeur « correcte » identifiable sans critère externe ; aucune étiquette n'a été utilisée pour la choisir.

$\Sigma$ : matrice de covariance ; $w$ : direction (composante) ; $\lambda_k$ : valeur propre associée à la composante $k$. **Paramètres appris :** les composantes (vecteurs propres). **Hyperparamètre :** nombre de composantes `n_components`.

### Présentation
**Catégorie :** non supervisé, réduction non linéaire pour la visualisation. **Hypothèse clé :** les voisinages locaux doivent être conservés. **Étapes :** calculer des probabilités de voisinage $p_{ij}$ en haute dimension (gaussienne, réglée par la perplexité), initialiser une carte 2D (PCA), calculer $q_{ij}$ (loi de Student), minimiser la divergence KL par descente de gradient.

$$C=\sum_i \mathrm{KL}(P_i\|Q_i)=\sum_{i\ne j}p_{ij}\ln\frac{p_{ij}}{q_{ij}},\qquad q_{ij}\propto\big(1+\|y_i-y_j\|^2\big)^{-1}$$

$p_{ij}$ : similarité des points $i,j$ dans l'espace d'origine ; $q_{ij}$ : similarité dans la carte ; $y_i$ : position 2D. **Paramètres appris :** les coordonnées 2D. **Hyperparamètres :** `perplexity`, `learning_rate`, initialisation.

### Schéma original
**À dessiner vous-même et à insérer ici (photo ou figure) :** Deux espaces côte à côte : un nuage 3D (ou une grille de points à 13 dimensions résumée) et la carte 2D ; relier quelques voisins par des fils pour montrer qu'ils doivent rester proches. Représenter l'évolution de la carte (désordre → groupes). Choix illustré = divergence KL.

*Conseil : légendez chaque étape et écrivez sous le dessin le choix algorithmique illustré.*

```python
# --- C4 : perplexity 12 vs 30 (voisinages préservés) ---
from sklearn.manifold import trustworthiness
fig, axes = plt.subplots(1, 2, figsize=(10, 4))
for ax, perp in zip(axes, [12, 30]):
    ts = TSNE(n_components=2, perplexity=perp, learning_rate='auto', init='pca', random_state=GRAINE)
    proj = ts.fit_transform(X_clu)
    print(f'perplexity={perp} | trustworthiness (5 voisins) : {trustworthiness(X_clu, proj, n_neighbors=5):.3f} | KL finale : {ts.kl_divergence_:.3f}')
    ax.scatter(proj[:, 0], proj[:, 1], s=14); ax.set_title(f't-SNE perplexity={perp}')
plt.show()
```

### Contextes et choix
- **Marketing :** cartographier visuellement des segments de clients. **Gestion :** explorer des profils d'entreprises.
- **Données :** taille petite à moyenne (coût élevé). **Concurrent :** UMAP, ACP.
- **Mal adapté :** interpréter des distances entre groupes ou des tailles de groupes ; projeter de nouvelles données (pas de transformation réutilisable).
- **Métrique :** trustworthiness (voisinages préservés), KL (non comparable entre perplexités).

### Test de sensibilité et interprétation
**Résultats du code :** perplexity 12 → trustworthiness (5 voisins) 0,969, KL finale 0,550 ; perplexity 30 → 0,967, KL 0,369.

**Interprétation.** La conservation des voisinages est quasi identique (écart 0,002) : la structure globale ressemble pour les deux perplexités ; observez la figure pour comparer visuellement la forme des groupes. **Catégorie :** non supervisé, réduction non linéaire (graphe de voisinage). **Hypothèse clé :** les données reposent sur une variété de faible dimension. **Étapes :** construire un graphe flou des $k$ plus proches voisins, initialiser la carte 2D (spectrale), optimiser une entropie croisée entre graphe d'origine et graphe de la carte.

$$\min_Y\ \sum_{i\ne j}\Big[\mu_{ij}\ln\frac{\mu_{ij}}{\nu_{ij}}+(1-\mu_{ij})\ln\frac{1-\mu_{ij}}{1-\nu_{ij}}\Big]$$

$\mu_{ij}$ : poids d'arête dans l'espace d'origine ; $\nu_{ij}$ : poids dans la carte ; $Y$ : coordonnées 2D. **Paramètres appris :** coordonnées 2D. **Hyperparamètres :** `n_neighbors`, `min_dist`.

```python
# --- C4 : n_neighbors = 8 puis 30 ---
fig, axes = plt.subplots(1, 2, figsize=(10, 4))
for ax, nn in zip(axes, [8, 30]):
    proj = umap.UMAP(n_components=2, n_neighbors=nn, min_dist=.25, random_state=GRAINE).fit_transform(X_clu)
    print(f'n_neighbors={nn} | trustworthiness (5 voisins) : {trustworthiness(X_clu, proj, n_neighbors=5):.3f}')
    ax.scatter(proj[:, 0], proj[:, 1], s=14); ax.set_title(f'UMAP n_neighbors={nn}')
plt.show()
```

## 26. Epsilon-greedy

**Variante C — question spécifique :** Quel effet de epsilon sur la concentration des choix ?

**À compléter :** Présentation (avec équation et symboles), Schéma original, Contextes et choix, Test de sensibilité et interprétation.

> Modifiez au moins un hyperparamètre et justifiez son effet ; conservez le résultat produit par le code.


```python
# Exploration epsilon-greedy sur le même bandit simulé
rng = np.random.default_rng(GRAINE)
probabilites_simulees = np.array([.35, .55, .65, .45])
recompenses = np.zeros(4); nombre_tirages = np.zeros(4)
epsilon = .2
for etape in range(1500):
    bras = int(rng.integers(4)) if rng.random() < epsilon or not np.all(nombre_tirages) else int(np.argmax(recompenses / np.maximum(nombre_tirages, 1)))
    gain = float(rng.random() < probabilites_simulees[bras])
    recompenses[bras] += gain; nombre_tirages[bras] += 1
print('Fréquences de sélection :', nombre_tirages.astype(int))
print('Valeurs estimées :', np.round(recompenses / np.maximum(nombre_tirages, 1), 3))
```

### Présentation
**Catégorie :** apprentissage par renforcement (cas sans état), **ici simulé**. **Hypothèse clé :** arbitrer entre explorer (essayer au hasard) et exploiter (jouer le meilleur bras connu). **Étapes :** avec probabilité $\varepsilon$ choisir un bras au hasard ; sinon le bras de moyenne estimée maximale ; observer la récompense, mettre à jour.

$$a_t=\begin{cases}\text{bras aléatoire} & \text{avec probabilité }\varepsilon\\ \arg\max_a\hat\mu_a & \text{avec probabilité }1-\varepsilon\end{cases}$$

$a_t$ : bras choisi ; $\hat\mu_a$ : moyenne estimée du bras $a$. **Paramètres appris :** les moyennes $\hat\mu_a$. **Hyperparamètre :** $\varepsilon$.

### Schéma original
**À dessiner vous-même et à insérer ici (photo ou figure) :** Un arbre de décision à deux branches à chaque tirage : « hasard (ε) » ou « meilleur actuel (1−ε) », avec des histogrammes des choix par bras pour ε = 0, 0,05, 0,5. Choix illustré = règle de décision (politique).

*Conseil : légendez chaque étape et écrivez sous le dessin le choix algorithmique illustré.*

```python
# --- C4 : epsilon et concentration des choix (20 graines) ---
vraies = np.array([.35, .55, .65, .45])  # probabilités SIMULÉES (non réelles)
def epsilon_greedy(eps, graine, n=1500):
    r = np.random.default_rng(graine); g = np.zeros(4); t = np.zeros(4); total = 0
    for _ in range(n):
        b = int(r.integers(4)) if (r.random() < eps or not np.all(t)) else int(np.argmax(g / np.maximum(t, 1)))
        gain = float(r.random() < vraies[b]); g[b] += gain; t[b] += 1; total += gain
    return t / n, total
print('epsilon | part des tirages sur le meilleur bras (moy ± éc.) | récompense totale (moy ± éc.)')
for eps in [0.0, .05, .2, .5]:
    res = [epsilon_greedy(eps, s) for s in range(20)]
    part = np.array([r[0][2] for r in res]); tot = np.array([r[1] for r in res])
    print(f'{eps:>7} | {part.mean():.3f} ± {part.std():.3f} | {tot.mean():.0f} ± {tot.std():.0f}')
```

### Contextes et choix
- **E-commerce :** choisir en ligne la version d'une page qui convertit le mieux tout en continuant à tester. **Marketing :** allocation du budget entre canaux.
- **Données :** retours en flux continu. **Concurrent :** UCB, Thompson.
- **Mal adapté :** environnement qui change (ε constant ne s'adapte pas bien) ; ε = 0 se bloque.
- **Métrique :** récompense cumulée et part de tirages sur le meilleur bras, sur **plusieurs graines**.

### Test de sensibilité et interprétation
**Résultats du code (20 graines, 1500 tirages ; valeurs données en moyenne ± écart-type).** Base (ε = 0,2, graine 42) : fréquences de sélection [83 119 1215 83]. Part de tirages sur le meilleur bras et récompense totale : ε = 0 → 0,597 ± 0,485 et 865 ± 149 ; ε = 0,05 → 0,838 ± 0,268 et 938 ± 46 ; ε = 0,2 → 0,786 ± 0,112 et 914 ± 26 ; ε = 0,5 → 0,585 ± 0,087 et 857 ± 21.

**Interprétation.** Avec ε = 0 (purement glouton) l'écart-type énorme (0,485) montre que l'algorithme se **bloque** parfois sur un mauvais bras. Un ε faible concentre mieux les choix sur le meilleur bras et donne la meilleure récompense moyenne ; un ε élevé garde de la variabilité faible mais gaspille des essais en exploration aléatoire.

**Limite :** 938 contre 914 (ε = 0,05 contre 0,2) : l'écart est suggestif (environ 2 erreurs-types) mais pas décisif avec 20 graines.

## 27. Q-learning

**Variante C — question spécifique :** Étudiez les récompenses des épisodes, pas seulement Q moyen.

**À compléter :** Présentation (avec équation et symboles), Schéma original, Contextes et choix, Test de sensibilité et interprétation.

> Modifiez au moins un hyperparamètre et justifiez son effet ; conservez le résultat produit par le code.


```python
# Q-learning manuel — environnement CliffWalking (Gymnasium)
import gymnasium as gym  # v1 : la v0 est obsolète dans Gymnasium récent
env = gym.make('CliffWalking-v1')
q_table = np.zeros((env.observation_space.n, env.action_space.n))
rng = np.random.default_rng(GRAINE)
for episode in range(1100):
    etat, _ = env.reset(seed=GRAINE + episode)
    epsilon = max(.08, .8 * (1 - episode / 900))
    for pas in range(120):
        action = int(rng.integers(env.action_space.n)) if rng.random() < epsilon else int(np.argmax(q_table[etat]))
        suivant, recompense, termine, tronque, _ = env.step(action)
        cible = recompense + (0 if termine else .95 * np.max(q_table[suivant]))
        q_table[etat, action] += .15 * (cible - q_table[etat, action])
        etat = suivant
        if termine or tronque: break
print('Table Q :', q_table.shape, '| valeur moyenne :', round(q_table.mean(), 2))
env.close()
```

### Présentation
**Catégorie :** apprentissage par renforcement, hors politique (*off-policy*), tabulaire, sur l'environnement simulé **CliffWalking** (grille 4×12, falaise = −100, chaque pas = −1). **Entrées :** état (case) ; **sortie :** action (4 directions). **Hypothèse clé :** processus de décision markovien. **Étapes :** choisir une action (epsilon-greedy), observer $(r,s')$, mettre à jour $Q$ avec la **meilleure** action suivante, répéter.

$$Q(s,a)\leftarrow Q(s,a)+\alpha\big[r+\gamma\max_{a'}Q(s',a')-Q(s,a)\big]$$

$s,a$ : état, action ; $r$ : récompense ; $s'$ : état suivant ; $\alpha$ : taux d'apprentissage ; $\gamma$ : facteur d'actualisation. **Paramètres appris :** la table $Q$ ($48\times4$). **Hyperparamètres :** $\alpha,\gamma,\varepsilon$ et son décroissement, nombre d'épisodes.

### Schéma original
**À dessiner vous-même et à insérer ici (photo ou figure) :** Une grille 4×12 avec la falaise en bas, départ et but aux deux coins ; trois cadres : table $Q$ à zéro, flèches qui se forment au fil des épisodes, trajet final collé à la falaise. Légende : l'algorithme apprend la politique optimale même si l'exploration est risquée. Choix illustré = cible avec le max.

*Conseil : légendez chaque étape et écrivez sous le dessin le choix algorithmique illustré.*

**Outils communs RL (Q-learning et SARSA)** : entraînement sur plusieurs graines, avec évaluation sans exploration.

```python
# --- Outils communs Q-learning / SARSA (plusieurs graines, évaluation sans exploration) ---
import gymnasium as gym
def entrainer(algo, graine, episodes=1100, alpha=.15, gamma=.95, eps_min=.08):
    env = gym.make('CliffWalking-v1'); q = np.zeros((48, 4)); r = np.random.default_rng(graine)
    agir = lambda s, e: int(r.integers(4)) if r.random() < e else int(np.argmax(q[s]))
    retours, chutes = [], []
    for ep in range(episodes):
        s, _ = env.reset(seed=graine + ep); e = max(eps_min, .8 * (1 - ep / 900)); a = agir(s, e); G = 0; ch = 0
        for _ in range(120):
            if algo == 'q': a = agir(s, e)
            s2, rew, term, trunc, _ = env.step(a); G += rew; ch += int(rew == -100)
            a2 = agir(s2, e)
            cible = rew + (0 if term else gamma * (q[s2].max() if algo == 'q' else q[s2, a2]))
            q[s, a] += alpha * (cible - q[s, a]); s = s2; a = a2
            if term or trunc: break
        retours.append(G); chutes.append(ch)
    env.close(); return q, np.array(retours), np.array(chutes)
def evaluer(q, eps, graine, n=30):
    env = gym.make('CliffWalking-v1'); r = np.random.default_rng(graine + 999); res = []
    for i in range(n):
        s, _ = env.reset(seed=i); G = 0
        for _ in range(120):
            a = int(r.integers(4)) if r.random() < eps else int(np.argmax(q[s]))
            s, rew, term, trunc, _ = env.step(a); G += rew
            if term or trunc: break
        res.append(G)
    env.close(); return np.mean(res)
print('Outils RL prêts')
```

```python
# --- C4 : récompenses d'épisodes, 5 graines (Q-learning) ---
res = []
for s in range(5):
    q, ret, ch = entrainer('q', s)
    res.append((ret[:100].mean(), ret[-100:].mean(), ret[-100:].std(), ch[-200:].sum(), evaluer(q, 0, s), evaluer(q, .1, s)))
res = np.array(res)
print('Q-learning | moyennes sur 5 graines (± écart-type entre graines)')
for nom, j in [('retour moyen des 100 premiers épisodes', 0), ('retour moyen des 100 derniers (avec exploration)', 1),
               ('chutes dans la falaise sur les 200 derniers épisodes', 3), ('évaluation gloutonne (epsilon=0)', 4), ('évaluation avec epsilon=0.1', 5)]:
    print(f'  {nom} : {res[:, j].mean():.1f} ± {res[:, j].std():.1f}')
q, ret, ch = entrainer('q', 0)
plt.plot(np.convolve(ret, np.ones(20) / 20, mode='valid')); plt.ylim(-200, 0)
plt.xlabel('Épisode'); plt.ylabel('Retour (moyenne glissante 20)'); plt.title('Q-learning — CliffWalking (graine 0)'); plt.show()
```

### Contextes et choix
- **Logistique :** apprentissage d'itinéraires dans un graphe de petite taille (simulation). **Gestion de stocks :** politique de réapprovisionnement avec états discrets.
- **Données :** environnement simulable avec états/actions discrets. **Concurrent :** SARSA.
- **Mal adapté :** espaces d'états continus ou très grands (table trop grande) ; environnement où une exploration risquée coûte cher en réel.
- **Métrique :** **retour par épisode** (somme des récompenses), pas la moyenne de la table Q.

### Test de sensibilité et interprétation
**Résultats du code (5 graines, moyenne ± écart-type entre graines) :** retour moyen des 100 premiers épisodes −876,3 ± 22,1 ; des 100 derniers (avec exploration) −41,7 ± 4,6 ; chutes dans la falaise sur les 200 derniers épisodes 53,4 ± 7,7 ; évaluation gloutonne (ε = 0) −13,0 ± 0,0 ; évaluation avec ε = 0,1 : −45,7 ± 11,9. Valeur moyenne de la table Q dans la cellule de base : −11,22.

**Interprétation.** La moyenne de la table $Q$ ne dit pas si la politique est bonne ; ce sont les **retours d'épisodes** qui le montrent. Q-learning trouve le trajet optimal (−13, collé à la falaise), mais ce trajet est dangereux dès que l'on explore un peu : −45,7 avec ε = 0,1 et environ 53 chutes sur les 200 derniers épisodes d'entraînement.

**Limite :** environnement simulé et déterministe ; les résultats ne se transposent pas à un cas réel sans validation.

## 28. SARSA

**Variante C — question spécifique :** SARSA privilégie-t-il une trajectoire plus sûre sous exploration ?

**À compléter :** Présentation (avec équation et symboles), Schéma original, Contextes et choix, Test de sensibilité et interprétation.

> Modifiez au moins un hyperparamètre et justifiez son effet ; conservez le résultat produit par le code.


```python
# SARSA manuel — même environnement, politique epsilon-greedy suivie
import gymnasium as gym  # v1 : la v0 est obsolète dans Gymnasium récent
env = gym.make('CliffWalking-v1')
q_table = np.zeros((env.observation_space.n, env.action_space.n))
rng = np.random.default_rng(GRAINE)
def choisir_action(etat, epsilon):
    return int(rng.integers(env.action_space.n)) if rng.random() < epsilon else int(np.argmax(q_table[etat]))
for episode in range(1100):
    etat, _ = env.reset(seed=GRAINE + episode)
    epsilon = max(.08, .8 * (1 - episode / 900))
    action = choisir_action(etat, epsilon)
    for pas in range(120):
        suivant, recompense, termine, tronque, _ = env.step(action)
        action_suivante = choisir_action(suivant, epsilon)
        cible = recompense + (0 if termine else .95 * q_table[suivant, action_suivante])
        q_table[etat, action] += .15 * (cible - q_table[etat, action])
        etat, action = suivant, action_suivante
        if termine or tronque: break
print('Table Q SARSA :', q_table.shape, '| valeur moyenne :', round(q_table.mean(), 2))
env.close()
```

### Présentation
**Catégorie :** apprentissage par renforcement, **sur politique** (*on-policy*), tabulaire, même environnement. **Hypothèse clé :** on évalue la politique réellement suivie, exploration comprise. **Étapes :** choisir $a$ dans $s$, observer $(r,s')$, choisir $a'$ dans $s'$ avec la même politique, mettre à jour avec $Q(s',a')$ et non avec le max.

$$Q(s,a)\leftarrow Q(s,a)+\alpha\big[r+\gamma\,Q(s',a')-Q(s,a)\big]$$

$a'$ : action effectivement choisie dans $s'$ ; autres symboles comme pour Q-learning. **Paramètres appris :** la table $Q$. **Hyperparamètres :** $\alpha,\gamma,\varepsilon$, épisodes.

### Schéma original
**À dessiner vous-même et à insérer ici (photo ou figure) :** La même grille que Q-learning mais deux trajets : celui de Q-learning collé à la falaise et celui de SARSA qui passe par le haut (plus long, plus sûr). Comparer en bas les deux formules de mise à jour, en soulignant « max » contre « action choisie ». Choix illustré = politique suivie vs politique optimale.

*Conseil : légendez chaque étape et écrivez sous le dessin le choix algorithmique illustré.*

```python
# --- C4 : SARSA vs Q-learning, mêmes grilles et mêmes graines ---
tab = {}
for algo in ['q', 'sarsa']:
    tab[algo] = np.array([(lambda q, ret, ch: (ret[-100:].mean(), ch[-200:].sum(), evaluer(q, 0, s), evaluer(q, .1, s)))(*entrainer(algo, s)) for s in range(5)])
print('algorithme | retour 100 derniers | chutes (200 derniers) | éval. epsilon=0 | éval. epsilon=0.1')
for algo, nom in [('q', 'Q-learning'), ('sarsa', 'SARSA')]:
    t = tab[algo]; print(f'{nom:>10} | ' + ' | '.join(f'{t[:, j].mean():7.1f} ± {t[:, j].std():.1f}' for j in range(4)))
q_s, _, _ = entrainer('sarsa', 0); q_q, _, _ = entrainer('q', 0)
def trajet(q):
    env = gym.make('CliffWalking-v1'); s, _ = env.reset(seed=0); lignes = [s // 12]
    for _ in range(60):
        s, _, term, trunc, _ = env.step(int(np.argmax(q[s]))); lignes.append(s // 12)
        if term or trunc: break
    env.close(); return lignes
print('Rangée visitée (0 = haut, la plus sûre ; 2 = rangée collée à la falaise) — Q-learning :', trajet(q_q))
print('Rangée visitée — SARSA                                                         :', trajet(q_s))
```

- **Données :** environnement simulable ; états/actions discrets. **Concurrent :** Q-learning.

- **Métrique :** retour par épisode, nombre de chutes, évaluation avec et sans exploration, sur plusieurs graines.

## 29. DQN — cible conceptuelle

**Variante C — question spécifique :** Pourquoi la cible est-elle égale à la récompense en état terminal ?

**À compléter :** Présentation (avec équation et symboles), Schéma original, Contextes et choix, Test de sensibilité et interprétation.

> Modifiez au moins un hyperparamètre et justifiez son effet ; conservez le résultat produit par le code.


```python
# Cibles DQN : démonstration du calcul de Bellman, SANS réseau neuronal
recompenses = np.array([-1., -100., 0.])
termine = np.array([False, True, False])
meilleure_valeur_suivante = np.array([-3., -2., 5.])
gamma = .9
cibles = recompenses + gamma * (~termine) * meilleure_valeur_suivante
print('Cibles de Bellman calculées :', np.round(cibles, 2))
print('Ce calcul ne constitue pas un entraînement DQN.')
```

### Présentation
**Catégorie :** apprentissage par renforcement profond — ici **uniquement le calcul de la cible de Bellman**, sans réseau de neurones ni entraînement (ce n'est pas un DQN entraîné). **Entrées :** lot de transitions $(r,\text{terminé},\max_{a'}Q)$ ; **sortie :** la cible que le réseau devrait prédire. **Étapes :** pour chaque transition, cible = récompense + $\gamma\times$ meilleure valeur suivante, sauf si l'état est terminal.

$$y=r+\gamma\,(1-d)\,\max_{a'}Q_{\text{cible}}(s',a')$$

$y$ : cible ; $r$ : récompense ; $d\in\{0,1\}$ : 1 si l'état est terminal ; $\gamma$ : actualisation ; $Q_{\text{cible}}$ : réseau cible. **Paramètres appris (dans un vrai DQN) :** les poids du réseau ; **ici : aucun**. **Hyperparamètre :** $\gamma$.

### Schéma original
**À dessiner vous-même et à insérer ici (photo ou figure) :** Un schéma en deux blocs : réseau en ligne qui prédit $Q(s,a)$ et réseau cible qui fournit $\max Q(s',a')$ ; une flèche vers la cible $y$ et un trait de perte entre $y$ et la prédiction. Indiquer en rouge « non entraîné ici ». Choix illustré = fonction objectif (erreur de Bellman).

*Conseil : légendez chaque étape et écrivez sous le dessin le choix algorithmique illustré.*

```python
# --- C4 : cible terminale et rôle de gamma ---
for g in [0., .5, .9, .99]:
    print(f'gamma={g:>4} | cibles : {np.round(recompenses + g * (~termine) * meilleure_valeur_suivante, 2)}')
print('État terminal (2e élément) : la cible vaut toujours', recompenses[1], 'car il n\'y a plus d\'état futur : (~termine) annule le terme gamma*Q.')
```

### Contextes et choix
- **Logistique/robotique :** apprentissage par réseau sur grand espace d'états (images de capteurs). **Finance :** stratégies sur états continus (en simulation).
- **Données :** simulateur ou grands historiques de transitions. **Concurrent :** Q-learning tabulaire (petits espaces).
- **Mal adapté :** peu de données, besoin de garanties de sécurité.
- **Métrique :** retour moyen par épisode et sa variabilité sur plusieurs graines, jamais la perte seule.

### Test de sensibilité et interprétation
**Résultats du code :** cibles pour γ = 0 / 0,5 / 0,9 / 0,99 : [−1 ; −100 ; 0] / [−2,5 ; −100 ; 2,5] / [−3,7 ; −100 ; 4,5] / [−3,97 ; −100 ; 4,95].

**Interprétation.** Dans l'état terminal (2ᵉ transition), la cible vaut toujours la récompense (−100) : il n'y a **aucun futur** à actualiser, le terme $\gamma\max Q$ est supprimé par $(1-d)$. Si on l'ajoutait, on attribuerait une valeur future à un état où l'épisode est fini. Pour les états non terminaux, γ proche de 1 donne plus de poids à l'avenir (cibles plus extrêmes).

**Limite :** aucun apprentissage n'a eu lieu ; ces cibles ne mesurent la performance d'aucun agent.

```python
# K-means manuel sur projection ACP du jeu Wine (sans étiquettes)
from sklearn.decomposition import PCA
from sklearn.cluster import KMeans
valeurs = PCA(n_components=2, random_state=GRAINE).fit_transform(X_clu)
rng = np.random.default_rng(GRAINE)
centroides = valeurs[rng.choice(len(valeurs), size=3, replace=False)].copy()
for iteration in range(100):
    distances = ((valeurs[:, None, :] - centroides[None, :, :]) ** 2).sum(axis=2)
    groupes = distances.argmin(axis=1)
    nouveaux = np.array([valeurs[groupes == k].mean(axis=0) if np.any(groupes == k) else centroides[k] for k in range(3)])
    if np.allclose(nouveaux, centroides):
        break
    centroides = nouveaux
print('Itérations manuelles :', iteration + 1)
print('Inertie manuelle :', round(((valeurs - centroides[groupes]) ** 2).sum(), 2))
modele_reference = KMeans(n_clusters=3, n_init=10, random_state=GRAINE).fit(valeurs)
print('Inertie scikit-learn :', round(modele_reference.inertia_, 2))
# L'inertie peut différer si l'initialisation est différente.
```

## C7/C8 — Bandits (SIMULÉS)
Référence : tirage cyclique. Approches : epsilon-greedy (ε = 0,1) et UCB1, 50 graines. Probabilités de gain fixées à [0,35 ; 0,55 ; 0,65 ; 0,45].

```python
# C7 — Bandits SIMULÉS : tirage cyclique (référence) vs epsilon-greedy vs UCB1, 50 graines
def bandit(politique, graine, n=1500):
    r = np.random.default_rng(graine); g = np.zeros(4); t = np.zeros(4); regret = np.zeros(n); cum = 0
    for i in range(n):
        if politique == 'cyclique': b = i % 4
        elif politique == 'eps': b = int(r.integers(4)) if (r.random() < .1 or not np.all(t)) else int(np.argmax(g / np.maximum(t, 1)))
        else: b = i if i < 4 else int(np.argmax(g / np.maximum(t, 1) + np.sqrt(2 * np.log(i + 1) / np.maximum(t, 1))))
        if politique == 'ucb': b = b % 4 if i < 4 else b
        g[b] += float(r.random() < vraies[b]); t[b] += 1; cum += vraies.max() - vraies[b]; regret[i] = cum
    return regret
vraies = np.array([.35, .55, .65, .45])
fig, ax = plt.subplots(figsize=(6, 4))
for pol, nom in [('cyclique', 'Cyclique (référence)'), ('eps', 'epsilon-greedy (0.1)'), ('ucb', 'UCB1')]:
    R = np.array([bandit(pol, s) for s in range(50)])
    print(f'{nom:>22} | regret cumulé final : {R[:, -1].mean():.1f} ± {R[:, -1].std():.1f}')
    ax.plot(R.mean(axis=0), label=nom)
ax.set_xlabel('Tirage'); ax.set_ylabel('Regret cumulé moyen'); ax.set_title('Bandit simulé (50 graines)'); ax.legend(); plt.show()
```

**Analyse.** Regret cumulé final sur 1500 tirages : cyclique 225,0 ± 0,0 (aucune exploitation, regret fixé par les écarts) ; epsilon-greedy 39,1 ± 30,0 ; UCB1 85,1 ± 13,3. L'epsilon-greedy a le meilleur regret moyen mais le plus grand écart-type (il peut parfois se bloquer sur un mauvais bras) ; UCB1 est plus régulier mais explore davantage avec ce bonus d'exploration. Résultats **simulés**, liés au choix des probabilités et à l'horizon de 1500 tirages.

## C7/C8 — Renforcement (CliffWalking, SIMULÉ)
Référence : politique aléatoire. Approches : Q-learning et SARSA, 5 graines, évaluation **sans exploration**.

```python
# C7/C8 — Renforcement (CliffWalking, simulé) : politique aléatoire (référence) vs Q-learning vs SARSA, 5 graines
def retour_aleatoire(graine, n=30):
    env = gym.make('CliffWalking-v1'); r = np.random.default_rng(graine); out = []
    for i in range(n):
        s, _ = env.reset(seed=i); G = 0
        for _ in range(120):
            s, rew, term, trunc, _ = env.step(int(r.integers(4))); G += rew
            if term or trunc: break
        out.append(G)
    env.close(); return np.mean(out)
alea = [retour_aleatoire(s) for s in range(5)]
print(f'Politique aléatoire : retour moyen {np.mean(alea):.1f} ± {np.std(alea):.1f}')
courbes = {}
for algo, nom in [('q', 'Q-learning'), ('sarsa', 'SARSA')]:
    ev, cs = [], []
    for s in range(5):
        q, ret, ch = entrainer(algo, s); ev.append(evaluer(q, 0, s)); cs.append(np.convolve(ret, np.ones(20) / 20, mode='valid'))
    courbes[nom] = np.mean(cs, axis=0)
    print(f'{nom:>10} | évaluation sans exploration par graine : {[float(e) for e in ev]} | moyenne {np.mean(ev):.1f} ± {np.std(ev):.1f}')
fig, ax = plt.subplots(figsize=(6.5, 4))
for nom, c in courbes.items(): ax.plot(c, label=nom)
ax.set_ylim(-250, 0); ax.set_xlabel('Épisode'); ax.set_ylabel('Retour (moyenne glissante 20, 5 graines)'); ax.set_title('CliffWalking'); ax.legend(); plt.show()
print('Rappel : -13 est le retour optimal ; -120 signifie que l\'épisode a été tronqué à 120 pas sans atteindre le but.')
```

**Analyse.** Politique aléatoire : −1241,3 ± 61,0 (nombreuses chutes). Q-learning : −13,0 sur les 5 graines (trajet optimal). SARSA : −58,2 ± 50,5, soit −17 (trajet sûr) pour 3 graines et −120 (épisode tronqué sans atteindre le but) pour 2 graines : SARSA n'a pas convergé partout avec 1100 épisodes. Les courbes d'apprentissage moyennes montrent que Q-learning gagne plus vite mais subit davantage de chutes pendant l'entraînement (voir algorithme 28). La variabilité entre graines (écart-type de SARSA) est un résultat à part entière, pas un détail.

## Bilan des limites et origine des différences observées
- **Données :** petits échantillons (178 à 569 observations) ; les écarts de 0,01 à 0,02 en F1 ou de 0,5 en MAE sont inférieurs à la variabilité entre séparations.
- **Hypothèses :** les modèles linéaires suffisent sur Diabetes et Breast Cancer ; les hypothèses d'indépendance (Naive Bayes) ou de sphéricité (K-means) coûtent peu ou beaucoup selon la redondance des variables.

- **Hasard :** graine des forêts, du K-means, de SARSA, qui change visiblement le résultat sur certaines graines.
- **Protocole :** un seul découpage entraînement/test pour la plupart des tables ; une répétition sur 10 séparations n'a été faite que pour la classification.
- **Simulation :** bandit et CliffWalking ne sont pas des données réelles ; DQN n'est qu'un calcul de cible, pas un entraînement.
